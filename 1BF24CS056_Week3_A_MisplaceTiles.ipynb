```python
import heapq


GOAL_STATE = ((1, 2, 3), (8, 0, 4), (7, 6, 5))

class PuzzleNodeMisplaced:
    def __init__(self, state, parent=None, action=None, g=0):
        self.state = state
        self.parent = parent
        self.action = action
        self.g = g
        self.h = self.calculate_heuristic()
        self.f = self.g + self.h

    def __lt__(self, other):
        return self.f < other.f

    def calculate_heuristic(self):
        misplaced = 0
        for r in range(3):
            for c in range(3):
                val = self.state[r][c]
                if val != 0 and val != GOAL_STATE[r][c]:
                    misplaced += 1
        return misplaced

    def find_blank(self):
        for r in range(3):
            for c in range(3):
                if self.state[r][c] == 0:
                    return r, c

    def generate_children(self):
        children = []
        r, c = self.find_blank()
        moves = [(-1, 0, 'UP'), (1, 0, 'DOWN'), (0, -1, 'LEFT'), (0, 1, 'RIGHT')]
        for dr, dc, action in moves:
            nr, nc = r + dr, c + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                new_state = [list(row) for row in self.state]
                new_state[r][c], new_state[nr][nc] = new_state[nr][nc], new_state[r][c]
                child_state = tuple(tuple(row) for row in new_state)
                children.append(PuzzleNodeMisplaced(child_state, self, action, self.g + 1))
        return children

def solve_misplaced(initial_state):
    open_list = [PuzzleNodeMisplaced(initial_state)]
    closed_set = set()
    nodes_expanded = 0

    while open_list:
        current = heapq.heappop(open_list)
        nodes_expanded += 1

        if current.state == GOAL_STATE:
            path = []
            curr = current
            while curr.parent:
                path.append((curr.action, curr.state))
                curr = curr.parent
            return path[::-1], current.g, nodes_expanded

        closed_set.add(current.state)
        for child in current.generate_children():
            if child.state not in closed_set:
                heapq.heappush(open_list, child)
    return None, 0, nodes_expanded

def print_puzzle(state):
    for row in state:
        print(" ".join(map(str, row)))
    print()


initial_state = ((2, 8, 3), (1, 6, 4), (7, 0, 5))
path, cost, expanded = solve_misplaced(initial_state)

print("=== CASE 1: STEP-BY-STEP TRACE (MISPLACED TILES) ===\n")
print("Initial State:")
print_puzzle(initial_state)

for i, (action, state) in enumerate(path, 1):
    print(f"Step {i}: Move Blank {action}")
    print_puzzle(state)

print(f"Goal reached in {cost} moves!")
print(f"Total Nodes Expanded: {expanded}")
```
