```python
board = [" "] * 9


def display_board():
    print()
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---+---+---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---+---+---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print()


def check_winner(symbol):
    winning_positions = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)
    ]

    for a, b, c in winning_positions:
        if board[a] == board[b] == board[c] == symbol:
            return True
    return False


def board_full():
    return " " not in board


def agent_move():
    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            if check_winner("O"):
                return
            board[i] = " "

    for i in range(9):
        if board[i] == " ":
            board[i] = "X"
            if check_winner("X"):
                board[i] = "O"
                return
            board[i] = " "

    if board[4] == " ":
        board[4] = "O"
        return

    for i in [0, 2, 6, 8]:
        if board[i] == " ":
            board[i] = "O"
            return

    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            return


print("TIC-TAC-TOE")
print("You are X, Agent is O")

while True:
    display_board()

    while True:
        try:
            position = int(input("Enter your position (1-9): ")) - 1

            if position < 0 or position > 8:
                print("Enter a number between 1 and 9.")
            elif board[position] != " ":
                print("Position already occupied.")
            else:
                board[position] = "X"
                break
        except ValueError:
            print("Please enter a valid number.")

    if check_winner("X"):
        display_board()
        print("Player wins!")
        break

    if board_full():
        display_board()
        print("Game Draw!")
        break

    agent_move()

    if check_winner("O"):
        display_board()
        print("Agent wins!")
        break

    if board_full():
        display_board()
        print("Game Draw!")
        break
```

```python
class VacuumCleanerAgent:

    def __init__(self, rooms):
        self.rooms = rooms
        self.position = 0

    def display(self):
        print("\nCurrent Environment:")
        for i, room in enumerate(self.rooms):
            status = "Dirty" if room == 1 else "Clean"
            location = " <- Vacuum" if i == self.position else ""
            print(f"Room {i + 1}: {status}{location}")

    def simple_reflex(self):
        if self.rooms[self.position] == 1:
            print(f"Vacuum cleans Room {self.position + 1}")
            self.rooms[self.position] = 0
            return True
        return False

    def goal_based(self):
        if all(room == 0 for room in self.rooms):
            return True

        for i in range(len(self.rooms)):
            if self.rooms[i] == 1:
                print(f"Vacuum moves from Room {self.position + 1} to Room {i + 1}")
                self.position = i
                return False

        return True

    def run(self):
        while True:
            self.display()

            if self.simple_reflex():
                continue

            if self.goal_based():
                print("\nGoal achieved!")
                print("All rooms are clean.")
                break


n = int(input("Enter number of rooms: "))

rooms = []

for i in range(n):
    while True:
        status = input(f"Is Room {i + 1} Dirty or Clean? ").lower()

        if status == "dirty":
            rooms.append(1)
            break
        elif status == "clean":
            rooms.append(0)
            break
        else:
            print("Please enter Dirty or Clean.")

agent = VacuumCleanerAgent(rooms)
agent.run()
```
