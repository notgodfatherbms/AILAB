```python
import random


def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])
    print()


def get_neighbors(state):
    neighbors = []

    zero = state.index(0)
    row, col = divmod(zero, 3)

    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_zero = new_row * 3 + new_col

            new_state = state.copy()

            new_state[zero], new_state[new_zero] = (
                new_state[new_zero],
                new_state[zero]
            )

            neighbors.append(new_state)

    return neighbors


def random_initial_state(goal, moves=10):
    state = goal.copy()
    previous = None

    for _ in range(moves):
        neighbors = get_neighbors(state)

        if previous is not None:
            filtered = [n for n in neighbors if n != previous]

            if filtered:
                neighbors = filtered

        previous = state
        state = random.choice(neighbors)

    return state


def depth_limited_search(
        state,
        goal,
        depth,
        path,
        path_set):

    if state == goal:
        return path

    if depth == 0:
        return None

    state_tuple = tuple(state)
    path_set.add(state_tuple)

    for neighbor in get_neighbors(state):

        neighbor_tuple = tuple(neighbor)

        if neighbor_tuple not in path_set:

            result = depth_limited_search(
                neighbor,
                goal,
                depth - 1,
                path + [neighbor],
                path_set
            )

            if result is not None:
                return result

    path_set.remove(state_tuple)

    return None


def ids(initial, goal):

    depth = 0

    while True:

        path_set = set()

        solution = depth_limited_search(
            initial,
            goal,
            depth,
            [initial],
            path_set
        )

        if solution is not None:
            return solution

        depth += 1



goal = [
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
]

initial = random_initial_state(goal, 10)

print("Initial State:")
print_puzzle(initial)

print("Goal State:")
print_puzzle(goal)

solution = ids(initial, goal)

print("IDS Solution:")

if solution:
    for i, state in enumerate(solution):
        print("Step", i)
        print_puzzle(state)
else:
    print("Solution not found.")
```

```python
import random

goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]


def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])
    print()


def get_neighbors(state):
    neighbors = []

    zero = state.index(0)
    row, col = divmod(zero, 3)

    moves = [
        (-1, 0),  # Up
        (1, 0),   # Down
        (0, -1),  # Left
        (0, 1)    # Right
    ]

    for dr, dc in moves:
        r = row + dr
        c = col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            new_zero = r * 3 + c

            new_state = state.copy()

            new_state[zero], new_state[new_zero] = (
                new_state[new_zero],
                new_state[zero]
            )

            neighbors.append(new_state)

    return neighbors


def random_initial_state(goal, moves=10):
    state = goal.copy()
    previous = None

    for _ in range(moves):
        neighbors = get_neighbors(state)

        if previous is not None:
            neighbors = [n for n in neighbors if n != previous]

        previous = state
        state = random.choice(neighbors)

    return state


def dfs(initial, goal):

    # DFS uses a stack
    stack = [(initial, [initial])]
    visited = set()

    while stack:

        state, path = stack.pop()

        if state == goal:
            return path

        state_tuple = tuple(state)

        if state_tuple in visited:
            continue

        visited.add(state_tuple)

        for neighbor in get_neighbors(state):

            if tuple(neighbor) not in visited:
                stack.append(
                    (neighbor, path + [neighbor])
                )

    return None


# Random initial state
initial = random_initial_state(goal, 10)

# DFS
solution = dfs(initial, goal)

# Print only the final path
if solution:

    for step, state in enumerate(solution):
        print("Step", step)
        print_puzzle(state)

    print("Total moves:", len(solution) - 1)

else:
    print("No solution")
```
