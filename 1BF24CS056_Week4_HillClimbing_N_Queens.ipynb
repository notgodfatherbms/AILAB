```python
import random

def calculate_cost(state):
    n = len(state)
    cost = 0
    for i in range(n):
        for j in range(i + 1, n):
            # Same row
            if state[i] == state[j]:
                cost += 1
            # Same diagonal
            elif abs(state[i] - state[j]) == abs(i - j):
                cost += 1
    return cost

def generate_neighbors(state):
    n = len(state)
    neighbors = []
    for col in range(n):
        for row in range(1, n + 1):
            if row != state[col]:
                neighbor = state.copy()
                neighbor[col] = row
                neighbors.append(neighbor)
    return neighbors

def hill_climbing(initial_state):
    current = initial_state.copy()
    while True:
        current_cost = calculate_cost(current)
        print("Current State:", current, "Cost:", current_cost)

        if current_cost == 0:
            print("Goal State Reached!\n")
            return current

        neighbors = generate_neighbors(current)
        best_neighbor = current
        best_cost = current_cost

        for neighbor in neighbors:
            cost = calculate_cost(neighbor)
            if cost < best_cost:
                best_neighbor = neighbor
                best_cost = cost

        # Local minimum check: no random restart
        if best_neighbor == current:
            print("Local minimum reached. Search stopped without restart.\n")
            return None

        current = best_neighbor

# Interactively ask the user for the number of queens
try:
    user_input = input("Enter the number of queens (e.g. 4, 8, 10): ")
    n = int(user_input)
    if n < 1:
        raise ValueError("Number of queens must be at least 1.")
except ValueError as e:
    print(f"Invalid input: {e}. Defaulting to N = 8.")
    n = 8

initial_random_state = [random.randint(1, n) for _ in range(n)]
print(f"\nGenerated Random Initial State for {n}-Queens:", initial_random_state)
solution = hill_climbing(initial_random_state)
if solution:
    print("Success! Solution:", solution)
else:
    print("Search failed to find a solution because it got stuck in a local minimum.")
```
